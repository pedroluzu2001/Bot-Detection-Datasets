```python
df = pd.read_csv("C:\\Users\\Pedroluzuriaga\\OneDrive - KARNATAKA SA\\Escritorio\\Bots_actualizado\\tweeets_no_sentiment.csv")

print("Shape:", df.shape)
df.head()
```

```python
def clean_text(text):
    if pd.isna(text):
        return ""
    text = text.strip()
    if len(text) < 5:
        return ""
    return text

df["clean_content"] = df["content"].apply(clean_text)
```

```python
INTENT_LABELS = [
    "attack",
    "defense",
    "propaganda",
    "mockery",
    "call to action",
    "neutral"
]
```

```python
classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli",
    device=DEVICE
)
```

```python
sample_text = df.loc[df["clean_content"] != "", "clean_content"].iloc[0]

classifier(
    sample_text,
    INTENT_LABELS,
    multi_label=False
)
```

```python
def infer_intent(text):
    if not isinstance(text, str) or text.strip() == "":
        return {label: 0.0 for label in INTENT_LABELS}

    result = classifier(
        text,
        INTENT_LABELS,
        multi_label=False
    )

    return dict(zip(result["labels"], result["scores"]))
```
